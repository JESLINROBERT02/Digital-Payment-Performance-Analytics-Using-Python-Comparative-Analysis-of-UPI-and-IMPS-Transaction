**Digital Payment Performance Analytics Using Python: Comparative Analysis of UPI and IMPS Transactions in India**

**OBJECTIVE:**


1.  To analyze the monthly transaction performance of UPI and IMPS in India from 2016 to 2026 using Python-based data analytics techniques.

1.   To compare the transaction volume and transaction value of UPI and IMPS in order to evaluate their growth trends and performance over the study period.

2.   To examine the year-wise and quarter-wise transaction performance of UPI and IMPS to identify seasonal trends and growth patterns.


2.  To examine the comparative growth, market performance and transaction behaviour of UPI and IMPS through statistical and graphical analysis.

1.   To provide data-driven insights and recommendations on the adoption and perfomrance of digital payment systems in India based on the comparative analysis of UPI and IMPS transactions





**DATASET INFORMATION**:

SOURCE: National Payments Corporation of India (NPCI) - Official Product Statistics for Unifited Payments Interface (UPI) and Immediate Payment Service (IMPS).

DATASET LINK: https://www.npci.org.in/product/upi/product-statistics and https://www.npci.org.in/product/imps/product-statistics

Location: India

Timeline: Monthly data collected Financial Year (April 2016 to March 2026)

Domain: Digital Payments / Banking & Financial Services / Financial Technology (FinTech)

BUSINESS PROBLEM:

The Rapid growth of digital payment systems has significantly transformed financial transactions in India. UPI and IMPS are two major payment platforms, but their transaction performance and growth patterns differ over time. Understanding these differences is important for financial institutions and policymakers to make informed decisions. However, there is limited comparative analysis using official transaction data. Therefore, this project uses Python to analyze and compare the perfomance of UPI and IMPS transactions in india.

**Data Transformation**

```python
# Convert Transaction Volume and Value Columns

# UPI

upi["Volume (In Mn.)"] = (
    upi["Volume (In Mn.)"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)


upi["Value (In Cr.)"] = (
    upi["Value (In Cr.)"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)


# IMPS

imps["Volume (In Mn.)"] = (
    imps["Volume (In Mn.)"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)


imps["Value (In Cr.)"] = (
    imps["Value (In Cr.)"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```

```python
# Rename Columns

upi.rename(columns={
    "Volume (In Mn.)": "Transaction_Volume_Mn",
    "Value (In Cr.)": "Transaction_Value_Cr"
}, inplace=True)


imps.rename(columns={
    "Volume (In Mn.)": "Transaction_Volume_Mn",
    "Value (In Cr.)": "Transaction_Value_Cr"
}, inplace=True)


# Display Column Names

print("UPI Columns:")
print(upi.columns)

print("\nIMPS Columns:")
print(imps.columns)
```

```python
# Convert Month Column

upi["Month"] = pd.to_datetime(
    upi["Month"],
    errors="coerce"
)


imps["Month"] = pd.to_datetime(
    imps["Month"],
    errors="coerce"
)


# Check Data Types

print(upi.dtypes)

print(imps.dtypes)
```

```python
# Sort Data by Month

upi = upi.sort_values("Month").reset_index(drop=True)

imps = imps.sort_values("Month").reset_index(drop=True)
```

**FEATURE ENGINEERING **

```python
# Add Payment Type Column

upi["Payment_Type"] = "UPI"

imps["Payment_Type"] = "IMPS"
```

```python
# Create Financial Year Function

def financial_year(date):
    if date.month >= 4:
        return str(date.year) + "-" + str(date.year + 1)
    else:
        return str(date.year - 1) + "-" + str(date.year)


# Apply Financial Year

upi["Financial_Year"] = upi["Month"].apply(financial_year)

imps["Financial_Year"] = imps["Month"].apply(financial_year)
```

```python
# Calculate Transaction Growth Percentage

upi["Transaction_Growth_%"] = (
    upi["Transaction_Volume_Mn"]
    .pct_change()
    * 100
)


imps["Transaction_Growth_%"] = (
    imps["Transaction_Volume_Mn"]
    .pct_change()
    * 100
)
```

```python
# Calculate Average Transaction Value

upi["Average_Transaction_Value"] = (
    upi["Transaction_Value_Cr"] /
    upi["Transaction_Volume_Mn"]
)


imps["Average_Transaction_Value"] = (
    imps["Transaction_Value_Cr"] /
    imps["Transaction_Volume_Mn"]
)
```

```python
# Create Quarter Feature

upi["Quarter"] = upi["Month"].dt.quarter

imps["Quarter"] = imps["Month"].dt.quarter
```

**STATISTICAL ANALYSIS**

```python
import pandas as pd

# Statistical Summary Function

def statistical_summary(df, payment_name):

    return pd.DataFrame({

        "Payment_Type": [payment_name],

        # Mean
        "Mean_Volume": [round(df["Transaction_Volume_Mn"].mean(),2)],
        "Mean_Value": [round(df["Transaction_Value_Cr"].mean(),2)],
        "Mean_Avg_Txn_Value": [round(df["Average_Transaction_Value"].mean(),2)],

        # Median
        "Median_Volume": [round(df["Transaction_Volume_Mn"].median(),2)],
        "Median_Value": [round(df["Transaction_Value_Cr"].median(),2)],
        "Median_Avg_Txn_Value": [round(df["Average_Transaction_Value"].median(),2)],

        # Mode
        "Mode_Volume": [round(df["Transaction_Volume_Mn"].mode()[0],2)],
        "Mode_Value": [round(df["Transaction_Value_Cr"].mode()[0],2)],
        "Mode_Avg_Txn_Value": [round(df["Average_Transaction_Value"].mode()[0],2)],

        # Variance
        "Variance_Volume": [round(df["Transaction_Volume_Mn"].var(),2)],
        "Variance_Value": [round(df["Transaction_Value_Cr"].var(),2)],
        "Variance_Avg_Txn_Value": [round(df["Average_Transaction_Value"].var(),2)],

        # Standard Deviation
        "Std_Volume": [round(df["Transaction_Volume_Mn"].std(),2)],
        "Std_Value": [round(df["Transaction_Value_Cr"].std(),2)],
        "Std_Avg_Txn_Value": [round(df["Average_Transaction_Value"].std(),2)],

        # Skewness
        "Skewness_Volume": [round(df["Transaction_Volume_Mn"].skew(),2)],
        "Skewness_Value": [round(df["Transaction_Value_Cr"].skew(),2)],
        "Skewness_Avg_Txn_Value": [round(df["Average_Transaction_Value"].skew(),2)],

        # Kurtosis
        "Kurtosis_Volume": [round(df["Transaction_Volume_Mn"].kurt(),2)],
        "Kurtosis_Value": [round(df["Transaction_Value_Cr"].kurt(),2)],
        "Kurtosis_Avg_Txn_Value": [round(df["Average_Transaction_Value"].kurt(),2)]

    })


# UPI Statistical Summary
upi_stats = statistical_summary(upi, "UPI")

# IMPS Statistical Summary
imps_stats = statistical_summary(imps, "IMPS")


# Final Statistical Analysis Table
statistical_analysis = pd.concat(
    [upi_stats, imps_stats],
    ignore_index=True
)


display(statistical_analysis)

```

**INTERPRETATION:**



1.  The mean transaction volume of UPI is 6079.21 million, which is significantly higher than IMPS (300.10 million), indicating the strong adoption and usage of UPI transactions.


1.   The mean transaction value of UPI is 892022.31 crore, whereas IMPS records 326926.98 crore, showing that UPI handles a larger overall transaction value.

1.   The average transaction value is higher for IMPS (1025.64) compared to UPI (175.74), indicating that individual IMPS transactions generally have higher value per transaction.
2.  The variance and standard deviation of transaction volume are higher for UPI (Variance: 47,485,510.69; Standard Deviation: 6890.97) compared to IMPS (Variance: 28,253.59; Standard Deviation: 168.09), showing greater variation in UPI transaction growth.


2.   The skewness values indicate that UPI transaction volume has positive skewness (0.95), while IMPS shows slight negative skewness (-0.21), representing different transaction distribution patterns.


2.   The kurtosis values for transaction volume are negative for both UPI (-0.45) and IMPS (-1.46), indicating a flatter distribution compared to a normal distribution.



1.   Overall, the statistical analysis shows that UPI dominates in transaction volume and total transaction value, while IMPS maintains a higher average transaction value per transaction.
